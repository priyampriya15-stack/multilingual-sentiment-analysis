# Exploratory Data Analysis: Multilingual Sentiment Analysis for Indian Languages

This notebook provides systematic data exploration for our 12 Indian languages + English + Hinglish sentiment benchmark dataset.

### Research Objectives
1. Quantify language representation and balance across 13 linguistic categories.
2. Inspect class distribution (Positive, Negative, Neutral) to identify class imbalance.
3. Analyze sentence length distributions across distinct Indic scripts.
4. Verify leakage-free stratified splitting across train, validation, and test sets.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Load dataset
df = pd.read_csv('../data/processed/multilingual_sentiment_dataset.csv')
print(f"Total dataset shape: {df.shape}")
df.head()
```

## 1. Class Distribution Across Languages

```python
print("Overall Sentiment Distribution:")
print(df['sentiment'].value_counts())

plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='sentiment', palette='viridis')
plt.title('Overall Sentiment Class Distribution')
plt.show()
```

## 2. Language Representation & Character Lengths

```python
df['char_length'] = df['text'].str.len()
plt.figure(figsize=(12, 5))
sns.boxplot(data=df, x='language', y='char_length', palette='Set2')
plt.title('Character Length Distribution Across Languages')
plt.xlabel('Language Code')
plt.ylabel('Character Length')
plt.show()
```

## 3. Train / Validation / Test Split Integrity

```python
split_counts = df.groupby(['split', 'sentiment']).size().unstack()
print(split_counts)

split_counts.plot(kind='bar', stacked=True, figsize=(9, 5), colormap='plasma')
plt.title('Stratified Split Composition')
plt.ylabel('Number of Samples')
plt.show()
```
